````python
# **Details:**

**Name:** Omkar Patil
**PRN:** 202401040125
**Div:** A2
**Roll No.:** 30

# **Title**

Intelligent Hospital Cleaning Robot Agent Using Python

# **Aim**

To design and implement an intelligent Hospital Cleaning Robot Agent using Python that can detect dirty hospital
rooms, prioritize the ICU for cleaning, move between rooms, manage battery consumption, clean dirty rooms, and generate a final cleaning report.

# **Experiment Objectives**

* To understand the concept of intelligent agents in Artificial Intelligence.
* To implement a simple Hospital Cleaning Robot Agent using Python.
* To represent the hospital environment using a Python dictionary.
* To detect dirty and clean hospital rooms.
* To give priority to the ICU when it is dirty.
* To move the cleaning robot between different rooms.
* To clean dirty hospital rooms.https://accounts.google.com/SignOutOptions?hl=en&continue=https://colab.research.google.com/drive/1Ouc8kv71e8eQvmNmK5zInXWa9PgSmeag&ec=GBRAqQM$0
* To manage the robot's battery during movement and cleaning.
* To recharge the robot when the battery level is low.
* To calculate the robot's score and number of movements.
* To display the final condition and cleaning statistics.

# **Experiment Outcome**

* To understand the working of an intelligent cleaning agent.
* To represent a hospital environment using Python data structures.
* To detect and clean dirty hospital rooms.
* To prioritize important areas such as the ICU.
* To move the robot between different rooms.
* To manage battery consumption and recharge when required.
* To calculate the robot's score and movements.
* To display the final state of the hospital and robot performance.

# **Theory**

### 1. **Introduction to Intelligent Agents**

An intelligent agent is a system that observes its environment and performs actions according to the current situation.

In this experiment, the Hospital Cleaning Robot observes the condition of different hospital rooms and takes appropriate actions such as moving, cleaning, and
recharging.

The robot maintains information about:

* Battery level
* Current position
* Cleaning score
* Number of movements

The robot makes decisions based on the cleanliness of the rooms and its battery level.

### 2. **Hospital Cleaning Robot Agent**

A Hospital Cleaning Robot is an example of an intelligent agent that performs cleaning tasks automatically in a hospital environment.

The hospital is represented using a Python dictionary. Each room has a value that represents its condition:

* `1` represents a dirty room.
* `0` represents a clean room.

For example:

```python
self.rooms = {
    "Reception": 1,
    "ICU": 1,
    "Ward": 0,
    "Pharmacy": 1,
    "Waiting Area": 0
}
```

The robot starts from the **Reception** and checks the hospital rooms for dirt. If the ICU is dirty, it is given the highest priority.

The robot can perform the following actions:

* **Move:** Move from one room to another.
* **Clean:** Clean a dirty room.
* **Recharge:** Recharge the battery when it becomes low.
* **Choose Room:** Select the next room to clean based on priority.
* **Generate Result:** Display the final cleaning status and robot performance.

### 3. **ICU Priority**

The main special feature of this agent is that it gives **first priority to the ICU**.

```python
if self.hospital.rooms["ICU"] == 1:
    return "ICU"
```

If the ICU is dirty, the robot selects it before other rooms. This makes the agent more suitable for a hospital environment where critical areas should be
cleaned first.

### 4. **Battery Management**

The robot starts with a battery level of **100%**.

* Moving to another room consumes **5%** battery.
* Cleaning a room consumes **15%** battery.
* When the battery falls below **30%**, the robot automatically recharges to **100%**.

This allows the robot to continue its cleaning task without stopping because of low battery.
````

````python
import random


class Hospital:
    def __init__(self):
        self.rooms = {
            "Reception": random.choice([0, 1]),
            "ICU": random.choice([0, 1]),
            "Ward": random.choice([0, 1]),
            "Pharmacy": random.choice([0, 1]),
            "Waiting Area": random.choice([0, 1])
        }


class CleaningRobot:
    def __init__(self, hospital):

        self.hospital = hospital
        self.position = "Reception"

        self.battery = 100
        self.score = 0
        self.moves = 0

        print("Hospital Condition:")
        print(self.hospital.rooms)

    def clean(self, room):

        print("\nCleaning:", room)

        self.hospital.rooms[room] = 0
        self.battery -= 15
        self.score += 10

        print("Cleaning completed.")
        print("Battery:", self.battery, "%")

    def move(self, room):

        print("\nMoving:", self.position, "->", room)

        self.position = room
        self.battery -= 5
        self.moves += 1
        self.score -= 1

    def choose_room(self):

        # ICU gets first priority
        if self.hospital.rooms["ICU"] == 1:
            return "ICU"

        # Select any other dirty room
        for room in self.hospital.rooms:
            if self.hospital.rooms[room] == 1:
                return room

        return None

    def recharge(self):

        print("\nBattery low!")
        print("Recharging...")

        self.battery = 100

        print("Battery:", self.battery, "%")

    def run(self):

        while True:

            room = self.choose_room()

            if room is None:
                print("\nAll hospital rooms are clean!")
                break

            # Recharge when battery is low
            if self.battery < 30:
                self.recharge()

            # Move to selected room
            if self.position != room:
                self.move(room)

            # Clean the room
            self.clean(room)

        self.result()

    def result(self):

        print("\n========== FINAL RESULT ==========")
        print("Hospital:", self.hospital.rooms)
        print("Current Position:", self.position)
        print("Battery:", self.battery, "%")
        print("Score:", self.score)
        print("Movements:", self.moves)


# Create hospital
hospital = Hospital()

# Create cleaning robot
robot = CleaningRobot(hospital)

# Start robot
robot.run()
````
